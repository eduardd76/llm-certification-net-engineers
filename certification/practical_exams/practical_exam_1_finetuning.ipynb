# Practical Exam 1 — Fine-Tune and Evaluate a Network LLM

**Professional Certification Practical Exam**

### Learning / assessment objectives
- Build a clean SFT experiment
- Train a LoRA/QLoRA adapter
- Evaluate against a held-out set
- Diagnose regressions

## Time budget
**3.5–4.5 hours**

## Scenario
Your company wants a local network assistant that:
- classifies incident type,
- produces concise troubleshooting plans,
- separates evidence from hypotheses,
- avoids unsafe first-step configuration changes.

You are given a small dataset. You must create a baseline, adapt a model, and evaluate it.

## Required deliverables
1. Baseline results
2. Data quality report
3. Training configuration
4. Fine-tuned adapter
5. Held-out evaluation
6. Regression analysis
7. Short model card
8. Recommendation: ship / do not ship

## Task A — Data audit (20 points)

Check:
- duplicates,
- leakage,
- class/task balance,
- unsupported claims,
- dynamic facts mistakenly placed in training,
- missing provenance.

Document at least three risks.

## Task B — Baseline (15 points)

Create at least 10 held-out prompts.
Measure:
- technical correctness,
- schema adherence,
- unsafe recommendation rate.

Do not tune against the final test prompts.

## Task C — Training (25 points)

Use LoRA or QLoRA.
Document:
- base model,
- rank,
- alpha,
- target modules,
- learning rate,
- epochs,
- max length,
- precision,
- hardware.

Explain why you chose them.

## Task D — Evaluation (25 points)

Compare base vs tuned model on exactly the same held-out prompts.
Include at least:
- one routing protocol case,
- one EVPN/VXLAN case,
- one MTU case,
- one general-knowledge sanity check,
- one safety case.

## Task E — Release decision (15 points)

Write:
- gains,
- regressions,
- known limitations,
- when RAG/tools are still required,
- whether you would deploy.

A correct "do not ship yet" can score higher than unjustified optimism.
