# Practical Exam 2 — Build a Bounded Agentic Network AI Workflow

**Professional Certification Practical Exam**

### Learning / assessment objectives
- Integrate RAG, read-only tools and validators
- Build a bounded agent loop
- Handle prompt injection and tool failures
- Design human approval boundaries

## Time budget
**3.5–4.5 hours**

## Scenario
An incident arrives:

> Users behind Leaf2 cannot reach an application after a network maintenance window.

Build a workflow that can:
- retrieve a runbook,
- query topology,
- query interface/BGP/VNI state,
- maintain structured incident state,
- propose next checks,
- stop when evidence is sufficient or limits are reached,
- propose but not execute a production change.

## Required deliverables
- architecture diagram,
- tool schemas,
- state model,
- agent/orchestrator loop,
- termination conditions,
- injection defense,
- validator,
- audit trace,
- example run.

## Required failure tests

Your workflow must handle:

1. Tool timeout.
2. Unknown device.
3. Retrieved document containing malicious instructions.
4. Model outputs malformed JSON.
5. Low-confidence diagnosis.
6. Proposed action outside authorized scope.

Each failure must produce a safe, observable outcome.

## Scoring

- Architecture separation: 20
- Tool/agent boundary: 15
- State/termination logic: 15
- Security / prompt-injection handling: 15
- Validation / approval gate: 15
- Observability: 10
- Example execution quality: 10

**Critical fail:** unrestricted state-changing tool use.
