# Troubleshooting Challenge Lab

**Professional Certification — Scenario Practice**

### Learning / assessment objectives
- Apply evidence-driven troubleshooting
- Separate observations from hypotheses
- Choose deterministic/tools/model mechanisms appropriately

## Instructions

For each challenge, produce:

1. **Observed evidence**
2. **Top hypotheses**
3. **Three next checks**
4. **What tool/data source you would query**
5. **What not to change yet**
6. **Confidence**
7. **Escalation / termination condition**

The grader values disciplined reasoning more than verbosity.

```python
import json, random
from pathlib import Path

ROOT = Path.cwd()
while not (ROOT/"certification/troubleshooting_challenges/challenge_bank.json").exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent
bank = json.load(open(ROOT/"certification/troubleshooting_challenges/challenge_bank.json", encoding="utf-8"))

def draw(n=5, seed=42):
    rng = random.Random(seed)
    for item in rng.sample(bank, min(n, len(bank))):
        print("\n", item["id"], item["domain"], item["difficulty"])
        print(item["scenario"])

draw()
```

## Scoring

Each scenario is 10 points:
- 2 evidence/hypothesis discipline
- 2 technical correctness
- 2 next checks
- 1 correct use of live tools/RAG/deterministic logic
- 1 safety / change restraint
- 1 confidence calibration
- 1 termination/escalation

A critical safety error caps the scenario at 4/10.
