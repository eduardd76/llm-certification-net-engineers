```python
QUESTIONS = [('What does DPO train from?', ['Only raw RFC text', 'Chosen/rejected preference pairs', 'Network graphs', 'Embeddings'], 1), ('What is a reward model?', ['A model that scores desirability', 'A tokenizer', 'A vector DB', 'A telemetry agent'], 0), ('Which is best suited to RLVR?', ['Subjective style preference', 'Exact subnet JSON', 'Brand tone', 'Creative explanation'], 1), ('What is reward hacking?', ['The model exploits the metric instead of the real intent', 'GPU failure', 'Dataset corruption only', 'Prompt caching'], 0), ('What is distillation?', ['Compressing logs', 'Training a student from a stronger teacher signal', 'Quantization only', 'RAG chunking'], 1)]

def run_quiz():
    score = 0
    for i, (q, options, answer) in enumerate(QUESTIONS, 1):
        print(f"\n{i}. {q}")
        for j, opt in enumerate(options):
            print(f"   {j}: {opt}")
        raw = input("Your answer (number): ").strip()
        try:
            choice = int(raw)
        except ValueError:
            choice = -1
        if choice == answer:
            print("Correct")
            score += 1
        else:
            print("Incorrect. Correct answer:", answer, "-", options[answer])
    pct = 100 * score / len(QUESTIONS)
    print(f"\nScore: {score}/{len(QUESTIONS)} = {pct:.1f}%")
    return pct

# Uncomment to take the quiz interactively.
# run_quiz()
```
