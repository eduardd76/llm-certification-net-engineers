```python
QUESTIONS = [('What is the best description of fine-tuning?', ['Adding current facts to the prompt', 'Updating model parameters using training data', 'Calling an external tool', 'Vector search'], 1), ('Which term describes the raw score before softmax?', ['Embedding', 'Logit', 'Gradient', 'Reward'], 1), ('Why is a context window not the same as memory?', ['It is only the tokens available to the current model call', 'It changes weights', 'It is a vector DB', 'It is always persistent'], 0), ('Which task is least suitable for an LLM?', ['Summarize incident notes', 'Explain BGP state', 'Exact CIDR containment check', 'Interpret a vague ticket'], 2), ('What does backpropagation compute?', ['Routing tables', 'Gradients', 'Embeddings only', 'Tool schemas'], 1)]

def run_quiz():
    score = 0
    for i, (q, options, answer) in enumerate(QUESTIONS, 1):
        print(f"\n{i}. {q}")
        for j, opt in enumerate(options):
            print(f"   {j}: {opt}")
        raw = input("Your answer (number): ").strip()
        try:
            choice = int(raw)
        except ValueError:
            choice = -1
        if choice == answer:
            print("Correct")
            score += 1
        else:
            print("Incorrect. Correct answer:", answer, "-", options[answer])
    pct = 100 * score / len(QUESTIONS)
    print(f"\nScore: {score}/{len(QUESTIONS)} = {pct:.1f}%")
    return pct

# Uncomment to take the quiz interactively.
# run_quiz()
```
