```python
QUESTIONS = [('What should the model do when current BGP state is required?', ['Guess', 'Call an authorized tool', 'Fine-tune', 'Use temperature 0'], 1), ('What is an agent?', ['Any chatbot', 'A bounded loop with state, actions/tools, observations and termination', 'Only a multi-model system', 'A vector DB'], 1), ('What is MCP mainly for?', ['Agent-to-agent delegation only', 'Connecting AI applications to tools/data/context providers', 'Fine-tuning', 'GPU scheduling'], 1), ('What is A2A mainly for?', ['Embedding search', 'Agent-to-agent interoperability', 'Quantization', 'Prompt templates'], 1), ('What is a termination condition?', ['A rule that stops an agent loop', 'A loss function', 'A tokenizer setting', 'A retrieval metric'], 0)]

def run_quiz():
    score = 0
    for i, (q, options, answer) in enumerate(QUESTIONS, 1):
        print(f"\n{i}. {q}")
        for j, opt in enumerate(options):
            print(f"   {j}: {opt}")
        raw = input("Your answer (number): ").strip()
        try:
            choice = int(raw)
        except ValueError:
            choice = -1
        if choice == answer:
            print("Correct")
            score += 1
        else:
            print("Incorrect. Correct answer:", answer, "-", options[answer])
    pct = 100 * score / len(QUESTIONS)
    print(f"\nScore: {score}/{len(QUESTIONS)} = {pct:.1f}%")
    return pct

# Uncomment to take the quiz interactively.
# run_quiz()
```
