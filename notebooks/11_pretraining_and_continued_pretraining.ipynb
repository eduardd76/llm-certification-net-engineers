## Pretraining

A causal LLM learns to predict subsequent tokens over enormous corpora. This creates broad language/world representations.
It does **not** inherently teach your desired incident schema or operational policy.

## Continued pretraining / DAPT

You can continue next-token training on a domain corpus—RFCs, design docs, validated runbooks—to shift the model toward domain language/statistics.

Use it when:
- the base model struggles with domain terminology/style,
- you have a large clean domain corpus,
- you can measure domain gains and general regressions.

Do **not** expect it to replace SFT for explicit instruction behavior.

```python
import requests, re
url = "https://www.rfc-editor.org/rfc/rfc4271.txt"
text = requests.get(url, timeout=30).text
print("BGP RFC chars:", len(text))
print(re.sub(r"\n{3,}","\n\n",text)[:1200])
```

## Document-level splitting matters

If chunks from the same RFC appear in both train and validation, validation perplexity can look artificially good.
For domain adaptation, split by document/source whenever possible.
