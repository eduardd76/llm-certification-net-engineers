## Two different interoperability problems

**MCP (Model Context Protocol)** standardizes how AI applications interact with external capabilities/context providers.
Think: **agent/application <-> tools/data**.

**A2A (Agent2Agent Protocol)** standardizes interoperable communication between independent agentic applications.
Think: **agent <-> agent**.

They can complement each other:
- NOC agent uses MCP-exposed NetBox/telemetry tools.
- NOC agent communicates with a SOC agent through A2A.

## Security questions to ask

Protocols do not remove trust problems:
- who authenticated?
- which tool/agent is authorized?
- what scopes?
- which inputs are untrusted?
- what is logged?
- can a retrieved document inject instructions?
- what happens if a peer lies or fails?
