## A model lifecycle

`raw corpus -> pretraining -> base model -> post-training -> instruct/chat model -> optional domain adaptation/alignment -> serving`

A **base model** mainly models token sequences. An **instruction-tuned** model is post-trained to follow requests.
A **reasoning model** is typically further optimized to spend inference computation on difficult problem solving.

Architecture labels are separate:
- **dense:** most parameters participate each token,
- **MoE:** routes tokens through selected experts,
- **VLM:** processes visual + language inputs,
- **SLM:** smaller model optimized for efficiency.

## Network model selection matrix

| Requirement | Likely priority |
|---|---|
| On-prem, low latency | smaller open-weights instruct model |
| Topology screenshot understanding | VLM |
| Bulk ticket classification | SLM/classifier |
| Complex multi-source diagnosis | stronger reasoning model + tools |
| Current device facts | any adequate model + retrieval/tools |
| 24/7 high volume | throughput, batching, cost, reliability |
