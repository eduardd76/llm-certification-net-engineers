### Learning goals
- Search the course terminology
- Create flashcards
- Test whether you can explain a term in networking language

```python
from pathlib import Path

def find_root():
    for p in [Path.cwd(), Path.cwd().parent, Path("/content/network_llm_engineering_course")]:
        if (p / "data" / "glossary.csv").exists():
            return p
    raise FileNotFoundError("Run from the extracted network_llm_engineering_course folder.")

ROOT = find_root()
DATA = ROOT / "data"
print("Course root:", ROOT)
```

```python
import pandas as pd
g = pd.read_csv(DATA/"glossary.csv")
print("Terms:", len(g))
display(g.head(10))
```

```python
def lookup(term):
    m = g[g["term"].str.contains(term, case=False, regex=False)]
    return m

display(lookup("LoRA"))
display(lookup("agent"))
```

```python
# Random flashcards.
sample = g.sample(10, random_state=42)
for t in sample["term"]:
    print("-", t)
```

### Self-test method

For each random term:
1. Explain it without jargon.
2. Give a networking example.
3. Say which layer it belongs to: data, model, training, retrieval, agent, serving, or assurance.
4. Give one common misconception.
