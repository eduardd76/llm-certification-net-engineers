## The end-to-end picture

A production Network AI system is rarely "an LLM". It is a **system**:

`data -> model -> retrieval/tools -> reasoning/orchestration -> verification -> policy -> deployment -> observability -> feedback`

This course teaches every layer, in dependency order.

### Six parts

1. **Foundations** — AI, ML, neural networks, transformers, tokens, inference.
2. **Knowledge** — prompting, embeddings, vector search, RAG, GraphRAG.
3. **Adaptation** — data engineering, continued pretraining, SFT, LoRA, QLoRA, evaluation.
4. **Alignment/Post-training** — preferences, reward models, DPO, RLHF, GRPO, RLVR, distillation.
5. **Agentic systems** — tools, structured outputs, agents, memory, MCP, A2A.
6. **Production Network AI** — AI NOC architecture, digital twins, serving, observability, safety, governance.

## The principle that prevents many bad architectures

**Model weights are not your CMDB, telemetry store, or routing table.**

Fine-tuning is best for relatively stable **behavior, skills, style, vocabulary, taxonomy, and response patterns**.
Dynamic facts — current interface state, inventory, configs, BGP sessions, incident tickets — belong in **retrieval and tools**.

## Course outcome

Your capstone is a small **Network Reasoning Assistant** that:
- retrieves network knowledge,
- calls mock/live tools,
- produces structured evidence-based diagnoses,
- can use a fine-tuned adapter,
- validates outputs before action,
- and is evaluated against held-out incidents.

The correct goal is not "make the LLM autonomous." The goal is **make the complete system reliable and measurable**.
