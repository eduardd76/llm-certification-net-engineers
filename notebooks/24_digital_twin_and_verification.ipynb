## Why a digital twin changes the AI architecture

A language model can propose a hypothesis or change. A digital twin can **execute/test** selected consequences in an isolated environment.

This creates a stronger loop:

`LLM proposal -> structured change -> policy check -> twin apply -> tests -> evidence -> approve/reject`

The twin is a verifier/environment, not a magical perfect copy of production.

```python
# Tiny deterministic reachability "twin" using a graph.
import networkx as nx
G = nx.Graph()
G.add_edges_from([
    ("Client","Leaf1"),
    ("Leaf1","Spine1"),
    ("Spine1","Leaf2"),
    ("Leaf2","Server"),
])
print("before:", nx.has_path(G,"Client","Server"))

# Simulate failed Spine1-Leaf2 link
G.remove_edge("Spine1","Leaf2")
print("after link failure:", nx.has_path(G,"Client","Server"))
```

## What needs modeling fidelity?

The twin only needs enough fidelity to verify the property under test.
For a routing policy change, you need routing behavior and topology—not necessarily full application stacks.
For a firewall/app vulnerability test, you may need representative services and versions.

**"One representative instance"** can be valid for some vulnerability behavior, but only if configuration, dependencies,
identity, policy, data flow, and version-specific behavior are equivalent enough for the test. Do not assume identical software version
automatically means identical operational risk.
