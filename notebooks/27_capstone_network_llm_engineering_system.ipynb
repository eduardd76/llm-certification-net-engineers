## Capstone problem

Build an assistant for six incident families:
- OSPF adjacency,
- BGP route/session,
- EVPN/VXLAN reachability,
- VLAN/STP,
- MTU/tunnel,
- DNS.

The assistant must never claim current state unless it came from supplied evidence or a tool.

## Required experiment matrix

Compare four systems on the same held-out incidents:

1. **Base + prompt**
2. **Base + RAG**
3. **Fine-tuned adapter**
4. **Fine-tuned adapter + RAG/tools + validators**

Measure:
- technical correctness,
- evidence/claim faithfulness,
- required checks covered,
- unsafe recommendation rate,
- schema validity,
- latency.

The expected lesson is not necessarily that #4 wins every metric. The lesson is to **measure which component solves which failure**.
