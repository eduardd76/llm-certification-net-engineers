## Three planes

**Knowledge plane:** RAG, KG, runbooks, historical incidents.
**Reasoning/action plane:** models, agents, tools, digital twin, validators.
**Assurance plane:** policy, approvals, evaluation, observability, audit, rollback.

A common architecture mistake is building only the reasoning plane.

## Model routing

Use the smallest adequate component:
- deterministic parser for exact work,
- SLM for high-volume classification,
- stronger LLM only for ambiguous reasoning,
- VLM only when visual input is required.

This often matters more economically than squeezing a few percent from a single giant model.
