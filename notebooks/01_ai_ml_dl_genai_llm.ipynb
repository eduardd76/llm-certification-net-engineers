## The hierarchy

**AI** is the umbrella. **ML** is one way to build AI. **Deep Learning** is a family of ML methods using deep neural networks.
**Generative AI** creates content. An **LLM** is a generative/self-supervised language model. An **agent** is a system that
uses a model in a loop with tools/state/actions.

An anomaly detector over interface telemetry may be AI/ML but **not** an LLM.
A BGP explainer can be an LLM application.
A NOC agent that queries telemetry, chooses checks, and escalates is an **agentic system**.

```python
use_cases = {
    "Predict interface failure from counters": "Machine Learning",
    "Cluster unknown traffic patterns": "Unsupervised ML",
    "Explain BGP route output": "LLM / Generative AI",
    "Generate a maintenance summary": "LLM / Generative AI",
    "Query NetBox, inspect telemetry, then propose diagnosis": "Agentic AI system",
    "Detect optical anomaly from time series": "ML / time-series model",
}
for task, category in use_cases.items():
    print(f"{task:58} -> {category}")
```

## A useful skeptical question

Before selecting an LLM, ask: **is language actually the hard part?**

If the core problem is forecasting packet loss, graph path computation, regex parsing, config diff, or subnet math, a deterministic
algorithm or classic ML model can be superior. LLMs are powerful because they bridge unstructured language and tools—not because
they should replace algorithms that already solve a problem exactly.
