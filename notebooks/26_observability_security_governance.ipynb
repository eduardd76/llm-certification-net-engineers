## Observe the system, not only the model

Capture:
- request/incident ID,
- model/version/adapter,
- prompt template version,
- retrieved source IDs,
- tool calls and arguments,
- tool results or hashes/redacted artifacts,
- model output,
- validator outcomes,
- human decision,
- latency/token counts/errors,
- final incident outcome.

This turns "the AI said..." into an auditable engineering system.

## Threats specific to agentic Network AI

- direct/indirect prompt injection,
- malicious ticket/runbook content,
- credential leakage,
- excessive tool permissions,
- unsafe shell/config generation,
- confused-deputy problems,
- poisoned training/retrieval data,
- untrusted agent peers,
- model supply-chain risk.

Controls should be deterministic where possible: identity, authorization, schemas, allowlists, network segmentation, sandboxing, approval.

## Release gates

A model/adapter version should not be promoted because "the loss looks good."
Require thresholds for:
- golden-set accuracy,
- safety scenarios,
- groundedness,
- schema validity,
- tool-call correctness,
- latency/cost,
- regression tests,
- rollback readiness.
